```python
#Gym Membership Fee
months = int(input("Enter membership months: "))
fee = float(input("Monthly fee: "))

total = months * fee

print("Total Membership Fee =", total)
```

```python
#Employee Working Hours
hours = float(input("Hours worked per day: "))
days = int(input("Working days: "))

total = hours * days

print("Total Working Hours =", total)
```

```python
#Fruit Shop Bill
apple = int(input("Number of apples: "))
banana = int(input("Number of bananas: "))

bill = (apple * 40) + (banana * 20)

print("Total Bill =", bill)
```

```python
#Package Weight Checker
weight = float(input("Enter package weight (kg): "))

if weight > 25:
    print("Heavy Package")
else:
    print("Normal Package")
```

```python
#Daily Water Intake
glasses = int(input("Enter glasses of water consumed: "))

if glasses >= 8:
    print("Good Hydration")
else:
    print("Drink More Water")
```

```python
#Classroom Attendance
present = int(input("Students Present: "))
total = int(input("Total Students: "))

attendance = (present / total) * 100

print("Attendance Percentage =", attendance)
```
