```python
#Brick Requirement
wall_length = 10
wall_height = 3
brick_area = 0.02
wall_area = wall_length * wall_height
bricks = wall_area / brick_area
print("Wall Area:", wall_area, "m²")
print("Required Bricks:", int(bricks))
```

```python
#Battery Charging Time
battery_capacity = 100
charger_power = 20
charging_efficiency = 0.85
charging_time = battery_capacity / (charger_power * charging_efficiency)
print("Battery Capacity:", battery_capacity, "kWh")
print("Charging Time:", round(charging_time, 2), "hours")
```

```python
#Heat Energy Calculation
mass = 50
specific_heat = 4.18
temperature_change = 30
heat = mass * specific_heat * temperature_change
print("Heat Required:", round(heat, 2), "kJ")
```

```python
#Braking Distance
speed = 20
friction = 0.7
gravity = 9.81
braking_distance = speed ** 2 / (2 * friction * gravity)
print("Braking Distance:", round(braking_distance, 2), "meters")
```

```python
#Engineering Unit Conversion
meters = 25
feet = meters * 3.28084
inches = meters * 39.3701
print("Length:", meters, "meters")
print("Length:", round(feet, 2), "feet")
print("Length:", round(inches, 2), "inches")
```

```python
#Engineering Project Budget
budget = 500000
materials = 180000
labor = 120000
equipment = 75000
transport = 45000
total_expense = materials + labor + equipment + transport
remaining = budget - total_expense
print("Total Expense: ₹", total_expense)
print("Remaining Budget: ₹", remaining)
if remaining >= 0:
    print("Project is within budget")
else:
    print("Budget exceeded")
```

```python
#Sensor Average Analysis
sensor_readings = [45, 48, 47, 52, 50, 49, 51, 46]
average = sum(sensor_readings) / len(sensor_readings)
maximum = max(sensor_readings)
minimum = min(sensor_readings)
print("Average:", round(average, 2))
print("Maximum:", maximum)
print("Minimum:", minimum)
```

```python
#Material Density
mass = 7850
volume = 1
density = mass / volume
print("Mass:", mass, "kg")
print("Volume:", volume, "m³")
print("Density:", density, "kg/m³")
```

```python
#Project Completion Tracker
tasks = {
    "Design": 100,
    "Planning": 100,
    "Construction": 70,
    "Testing": 40,
    "Documentation": 20
}
total_progress = sum(tasks.values()) / len(tasks)

print("Project Progress:", round(total_progress, 2), "%")

if total_progress >= 80:
    print("Project Status: Almost Complete")
elif total_progress >= 50:
    print("Project Status: In Progress")
else:
    print("Project Status: Initial Stage")
```

```python
#Best Machine Selection
machines = {
    "Machine A": {"efficiency": 88, "cost": 50000},
    "Machine B": {"efficiency": 94, "cost": 65000},
    "Machine C": {"efficiency": 91, "cost": 55000}
}
best_machine = max(
    machines,
    key=lambda x: machines[x]["efficiency"]
)
print("Best Machine:", best_machine)
print("Efficiency:", machines[best_machine]["efficiency"], "%")
print("Cost: ₹", machines[best_machine]["cost"])
```
