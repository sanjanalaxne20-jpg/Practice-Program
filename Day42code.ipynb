```python
#Laptop EMI
price = float(input("Enter laptop price: "))
months = int(input("Enter EMI months: "))

emi = price / months

print("Monthly EMI = ₹", round(emi, 2))
```

```python
#Milk Dairy Collection
liters = float(input("Enter milk quantity: "))
rate = float(input("Enter rate per liter: "))

amount = liters * rate

print("Amount = ₹", amount)
```

```python
#Employee Leave Balance
total_leave = int(input("Total leave: "))
used_leave = int(input("Used leave: "))

balance = total_leave - used_leave

print("Leave Balance =", balance)
```

```python
#Mobile Recharge Validity
validity = int(input("Enter validity days: "))

if validity >= 28:
    print("Valid Recharge")
else:
    print("Recharge Soon")
```
