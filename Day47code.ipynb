```python
#ATM PIN Verification
pin = input("Enter PIN: ")

if pin == "1234":
    print("Access Granted")
else:
    print("Invalid PIN")
```

```python
#Parcel Volume
length = float(input("Length: "))
width = float(input("Width: "))
height = float(input("Height: "))

volume = length * width * height

print("Parcel Volume =", volume)
```

```python
#Movie Rating
rating = float(input("Enter movie rating: "))

if rating >= 4:
    print("Excellent")
else:
    print("Average")
```

```python
#Bookstore Bill
books = int(input("Enter number of books: "))
price = float(input("Price per book: "))

bill = books * price

print("Book Bill = ₹", bill)
```

```python
#Student Age Category
age = int(input("Enter age: "))

if age < 13:
    print("Child")
elif age < 18:
    print("Teenager")
else:
    print("Adult")
```

```python
#Internet Bill
plan = float(input("Monthly Plan Amount: "))
tax = plan * 0.18

print("Total Bill =", plan + tax)
```

```python
#Electricity Saving
last = int(input("Last Month Units: "))
current = int(input("Current Month Units: "))

if current < last:
    print("Electricity Saved")
else:
    print("Consumption Increased")
```

```python
#Car Speed Checker
speed = int(input("Enter Speed: "))

if speed > 80:
    print("Overspeeding")
else:
    print("Safe Speed")
```

```python
#Scholarship Eligibility
marks = float(input("Enter Percentage: "))

if marks >= 85:
    print("Scholarship Eligible")
else:
    print("Not Eligible")
```

```python
#Fruit Discount
amount = float(input("Enter Bill Amount: "))

if amount > 1000:
    print("Discount = ₹100")
else:
    print("No Discount")
```
