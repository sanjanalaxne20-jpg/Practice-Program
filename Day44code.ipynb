```python
#Employee Attendance
present = int(input("Present days: "))
working = int(input("Working days: "))

attendance = (present / working) * 100

print("Attendance =", attendance, "%")
```

```python
#Grocery Store Bill
rice = float(input("Rice cost: "))
oil = float(input("Oil cost: "))
sugar = float(input("Sugar cost: "))

bill = rice + oil + sugar

print("Total Grocery Bill =", bill)
```

```python
#Internet Speed Check
speed = float(input("Enter internet speed (Mbps): "))

if speed >= 50:
    print("High Speed Internet")
else:
    print("Low Speed Internet")
```

```python
#Classroom Bench Requirement
students = int(input("Number of students: "))
bench_capacity = int(input("Students per bench: "))

benches = (students + bench_capacity - 1) // bench_capacity

print("Benches Required =", benches)
```

```python
#Petrol Consumption
distance = float(input("Distance traveled (km): "))
mileage = float(input("Mileage (km/l): "))

fuel = distance / mileage

print("Fuel Required =", fuel, "liters")
```

```python
#Restaurant Tip Calculator
bill = float(input("Enter bill amount: "))
tip = float(input("Enter tip percentage: "))

tip_amount = bill * tip / 100

print("Tip =", tip_amount)
print("Total Payment =", bill + tip_amount)
```

```python
#Phone Storage Checker
total = float(input("Total storage (GB): "))
used = float(input("Used storage (GB): "))

free = total - used

print("Available Storage =", free, "GB")
```

```python
#Warehouse Inventory
received = int(input("Items received: "))
sold = int(input("Items sold: "))

stock = received - sold

print("Remaining Stock =", stock)
```

```python
#Salary Deduction
salary = float(input("Salary: "))
deduction = float(input("Deduction: "))

net_salary = salary - deduction

print("Net Salary =", net_salary)
```
