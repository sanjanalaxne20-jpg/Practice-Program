```python
#CPU Temperature Monitoring
temperatures = [55, 62, 68, 75, 82, 88, 70]

for i, temp in enumerate(temperatures, start=1):

    if temp >= 85:
        status = "Critical"
    elif temp >= 75:
        status = "High"
    else:
        status = "Normal"

    print(
        "Reading", i,
        ":", temp, "°C ->", status
    )
```

```python
#Network Packet Loss
sent = 10000
received = 9850

lost = sent - received
loss_percentage = (lost / sent) * 100

print("Packets Sent:", sent)
print("Packets Lost:", lost)
print("Packet Loss:", loss_percentage, "%")

if loss_percentage > 2:
    print("Network Quality: Poor")
else:
    print("Network Quality: Good")
```

```python
#Mobile Sensor Data Analysis
accelerometer = {
    "X": [1.2, 1.5, 1.3, 1.8],
    "Y": [0.8, 0.9, 1.0, 0.7],
    "Z": [9.7, 9.8, 9.6, 9.9]
}

for axis, values in accelerometer.items():

    average = sum(values) / len(values)

    print(
        axis,
        "Average:",
        round(average, 2)
    )
```

```python
#Fire Alarm Detection
temperature = 92
smoke_level = 78

if temperature > 80 and smoke_level > 60:
    print("FIRE ALERT!")
elif temperature > 80:
    print("High Temperature Warning")
elif smoke_level > 60:
    print("Smoke Warning")
else:
    print("Environment Normal")
```

```python
#Drone Flight Time
battery = 5000
power_consumption = 250

flight_time = battery / power_consumption

print("Estimated Flight Time:",
      round(flight_time, 2), "hours")
```

```python
#Factory Overall Efficiency
departments = {
    "Assembly": 92,
    "Packaging": 87,
    "Testing": 95,
    "Quality": 90
}

overall = sum(departments.values()) / len(departments)

print("Department Efficiencies:")

for department, efficiency in departments.items():
    print(department, ":", efficiency, "%")

print("\nOverall Factory Efficiency:",
      round(overall, 2), "%")
```

```python
#Material Volume Optimization
length = 10
width = 5
height = 2

original_volume = length * width * height

reduction = 15 / 100
new_volume = original_volume * (1 - reduction)

saved_volume = original_volume - new_volume

print("Original Volume:", original_volume, "m³")
print("Optimized Volume:", new_volume, "m³")
print("Material Saved:", saved_volume, "m³")
```

```python
#Machine Downtime Cost
machines = {
    "Machine A": 4,
    "Machine B": 7,
    "Machine C": 3
}

cost_per_hour = 2500

total_downtime = sum(machines.values())
total_loss = total_downtime * cost_per_hour

print("Total Downtime:", total_downtime, "hours")
print("Production Loss: ₹", total_loss)
```

```python
#Server Capacity Planning
servers = {
    "Server A": 65,
    "Server B": 72,
    "Server C": 88,
    "Server D": 94
}

high_load = []

for server, usage in servers.items():

    if usage > 80:
        high_load.append(server)

print("Servers Requiring Attention:")

for server in high_load:
    print(server)
```

```python
#Automated Engineering Decision System
temperature = 82
vibration = 4.2
pressure = 95

risk_score = 0

if temperature > 80:
    risk_score += 1

if vibration > 3.5:
    risk_score += 1

if pressure > 90:
    risk_score += 1

print("Risk Score:", risk_score, "/ 3")

if risk_score == 0:
    print("Machine Status: Safe")
elif risk_score == 1:
    print("Machine Status: Monitor")
elif risk_score == 2:
    print("Machine Status: Warning")
else:
    print("Machine Status: Critical - Stop Machine")
```
