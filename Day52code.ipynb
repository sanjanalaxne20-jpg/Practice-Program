```python
#Digital Clock – AM/PM
hour = int(input("Enter hour (0-23): "))

if hour < 12:
    print("AM")
else:
    print("PM")
```

```python
#Password Strength
password = input("Enter password: ")

if len(password) < 6:
    print("Weak Password")
elif len(password) < 10:
    print("Medium Password")
else:
    print("Strong Password")
```

```python
#Username Generator
name = input("Enter your name: ")
year = input("Enter birth year: ")

username = name[:3].lower() + year

print("Generated Username =", username)
```

```python
#Number of Digits
number = int(input("Enter a number: "))

digits = len(str(abs(number)))

print("Number of Digits =", digits)
```

```python
#Reverse a Number
number = input("Enter a number: ")

reverse = number[::-1]

print("Reversed Number =", reverse)
```

```python
#Air Quality Category
aqi = int(input("Enter AQI value: "))

if aqi <= 50:
    print("Air Quality: Good")
elif aqi <= 100:
    print("Air Quality: Satisfactory")
elif aqi <= 200:
    print("Air Quality: Moderate")
elif aqi <= 300:
    print("Air Quality: Poor")
else:
    print("Air Quality: Very Poor")
```

```python
#Carbon Emission from Vehicle
distance = float(input("Enter distance travelled (km): "))
fuel = float(input("Enter fuel consumed (liters): "))
emission_factor = 2.31

emission = fuel * emission_factor

print("CO2 Emission =", round(emission, 2), "kg")
```

```python
#Waste Recycling Rate
total_waste = float(input("Enter total waste (kg): "))
recycled_waste = float(input("Enter recycled waste (kg): "))

rate = (recycled_waste / total_waste) * 100

print("Recycling Rate =", round(rate, 2), "%")
```

```python
#Water Quality – pH Classification
ph = float(input("Enter water pH: "))

if ph < 6.5:
    print("Water is Acidic")
elif ph <= 8.5:
    print("Water is within Normal Range")
else:
    print("Water is Alkaline")
```

```python
#Daily Water Consumption
people = int(input("Enter number of people: "))
water_per_person = float(input("Water used per person (liters): "))

total = people * water_per_person

print("Daily Water Consumption =", total, "liters")
```
