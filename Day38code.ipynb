```python
#Electricity Department – Bill
units = int(input("Enter electricity units: "))
rate = 8

bill = units * rate

print("Electricity Bill = ₹", bill)
```

```python
#Data Analyst – Sales Analysis
jan = float(input("January Sales: "))
feb = float(input("February Sales: "))
mar = float(input("March Sales: "))

total = jan + feb + mar
average = total / 3

print("Total Sales =", total)
print("Average Sales =", average)
```

```python
#Employee Management – Bonus
salary = float(input("Enter salary: "))

bonus = salary * 0.10

print("Bonus = ₹", bonus)
print("Total Salary = ₹", salary + bonus)
```

```python
#Airline – Baggage Fee
weight = float(input("Enter baggage weight (kg): "))

if weight > 20:
    fee = (weight - 20) * 100
    print("Extra Fee = ₹", fee)
else:
    print("No Extra Charge")
```

```python
#Pharmacy – Medicine Bill
medicine1 = float(input("Price of Medicine 1: "))
medicine2 = float(input("Price of Medicine 2: "))
medicine3 = float(input("Price of Medicine 3: "))

bill = medicine1 + medicine2 + medicine3

print("Medicine Bill = ₹", bill)
```

```python
#ATM Withdrawal
balance = float(input("Enter account balance: "))
withdraw = float(input("Enter withdrawal amount: "))

if withdraw <= balance:
    balance -= withdraw
    print("Remaining Balance =", balance)
else:
    print("Insufficient Balance")
```

```python
#Car Mileage
distance = float(input("Enter distance (km): "))
fuel = float(input("Enter fuel used (liters): "))

mileage = distance / fuel

print("Mileage =", mileage, "km/l")
```

```python
#Mobile Data Usage
used = float(input("Enter data used (GB): "))
limit = float(input("Enter data limit (GB): "))

remaining = limit - used

print("Remaining Data =", remaining, "GB")
```

```python
#Restaurant GST Bill
bill = float(input("Enter food bill: "))

gst = bill * 0.05
total = bill + gst

print("GST =", gst)
print("Total Bill =", total)
```

```python
#Movie Collection
tickets = int(input("Tickets sold: "))
price = float(input("Ticket price: "))

collection = tickets * price

print("Total Collection =", collection)
```
