```python
#Bus Fuel Requirement
distance = float(input("Enter journey distance: "))
mileage = float(input("Enter bus mileage: "))

fuel = distance / mileage

print("Fuel Required =", round(fuel, 2), "liters")
```

```python
#Warehouse Stock Alert
stock = int(input("Enter current stock: "))
minimum = int(input("Enter minimum stock level: "))

if stock <= minimum:
    print("Stock Alert: Reorder Required")
else:
    print("Stock Level is Sufficient")
```

```python
#Flight Baggage Check
weight = float(input("Enter baggage weight: "))

if weight <= 15:
    print("No Extra Charge")
else:
    extra = weight - 15
    charge = extra * 500
    print("Extra Baggage Charge = ₹", charge)
```

```python
#Gym Calorie Goal
calories = int(input("Enter calories burned: "))
goal = int(input("Enter daily calorie goal: "))

if calories >= goal:
    print("Goal Achieved")
else:
    remaining = goal - calories
    print("Calories Remaining =", remaining)
```

```python
#Hospital Room Selection
room = input("Enter room type (general/private/icu): ")

if room == "general":
    cost = 1500
elif room == "private":
    cost = 3000
elif room == "icu":
    cost = 7000
else:
    cost = 0

print("Daily Room Charge = ₹", cost)
```

```python
#Parking Slot Availability
total_slots = int(input("Enter total parking slots: "))
occupied = int(input("Enter occupied slots: "))

available = total_slots - occupied

print("Available Slots =", available)

if available == 0:
    print("Parking Full")
```

```python
#Online Course Progress
total_modules = int(input("Enter total modules: "))
completed = int(input("Enter completed modules: "))

progress = (completed / total_modules) * 100

print("Course Progress =", round(progress, 2), "%")
```

```python
#Employee Overtime Checker
hours = float(input("Enter working hours: "))

if hours > 8:
    overtime = hours - 8
    print("Overtime Hours =", overtime)
else:
    print("No Overtime")
```

```python
#Grocery Stock Check
required = int(input("Enter required quantity: "))
available = int(input("Enter available quantity: "))

if available >= required:
    print("Order can be fulfilled")
else:
    shortage = required - available
    print("Shortage =", shortage)
```

```python
#YouTube Video Engagement
views = int(input("Enter views: "))
likes = int(input("Enter likes: "))

like_rate = (likes / views) * 100

print("Like Rate =", round(like_rate, 2), "%")
```
