```python
#Solar Plant Capacity Factor
actual = float(input("Enter actual energy generated (kWh): "))
capacity = float(input("Enter plant capacity (kW): "))
hours = float(input("Enter operating period (hours): "))
maximum = capacity * hours
factor = (actual / maximum) * 100
print("Capacity Factor =", round(factor, 2), "%")
```

```python
#Rainfall Runoff
rainfall = float(input("Enter rainfall (mm): "))
area = float(input("Enter catchment area (m²): "))
runoff_coefficient = float(input("Enter runoff coefficient: "))
runoff = rainfall * area * runoff_coefficient
print("Estimated Runoff =", runoff, "liters")
```

```python
#Renewable Energy Percentage
total = float(input("Enter total energy consumption (kWh): "))
renewable = float(input("Enter renewable energy used (kWh): "))
percentage = (renewable / total) * 100
print("Renewable Energy Share =", round(percentage, 2), "%")
```

```python
#Carbon Offset Requirement
emission = float(input("Enter carbon emission (tons): "))
offset = float(input("Enter carbon already offset (tons): "))
remaining = emission - offset
if remaining > 0:
    print("Additional Offset Required =", remaining, "tons")
else:
    print("Carbon Offset Target Achieved")
```

```python
#Air Pollution Monitoring
pm25 = float(input("Enter PM2.5 concentration (µg/m³): "))
if pm25 <= 30:
    print("Pollution Level: Low")
elif pm25 <= 60:
    print("Pollution Level: Moderate")
else:
    print("Pollution Level: High")
```

```python
#Waste Generation per Person
waste = float(input("Enter total waste generated (kg): "))
population = int(input("Enter population: "))
per_person = waste / population
print("Waste per Person =", round(per_person, 2), "kg")
```

```python
#Energy Intensity
energy = float(input("Enter energy consumed (kWh): "))
production = float(input("Enter production (units): "))
intensity = energy / production
print("Energy Intensity =", round(intensity, 2), "kWh/unit")
```

```python
#Tree Plantation Requirement
target = int(input("Enter target number of surviving trees: "))
survival_rate = float(input("Enter expected survival rate (%): "))
trees_to_plant = target / (survival_rate / 100)
print("Trees to be Planted =", round(trees_to_plant))
```

```python
#Water Treatment Recovery
input_water = float(input("Enter water entering treatment plant (liters): "))
recovery_rate = float(input("Enter recovery rate (%): "))
recovered = input_water * recovery_rate / 100
print("Treated Water Recovered =", recovered, "liters")
```

```python
#Environmental Incident Severity
impact = int(input("Enter environmental impact score (1-10): "))
if impact <= 3:
    print("Severity: Low")
elif impact <= 6:
    print("Severity: Medium")
elif impact <= 8:
    print("Severity: High")
else:
    print("Severity: Critical")
```
