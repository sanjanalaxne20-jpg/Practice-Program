```python
#Reverse Words
def reverse_words(text):
    print(" ".join(text.split()[::-1]))

s = input("Enter sentence: ")
reverse_words(s)
```

```python
#Count Uppercase Letters
def upper_count(text):
    count = 0
    for ch in text:
        if ch.isupper():
            count += 1
    print(count)

s = input("Enter text: ")
upper_count(s)
```

```python
#Check Binary Number
def binary(num):
    if set(num) <= {"0", "1"}:
        print("Valid Binary")
    else:
        print("Invalid Binary")

n = input("Enter binary: ")
binary(n)
```

```python
#Decimal to Binary
def decimal_binary(num):
    print(bin(num)[2:])

n = int(input("Enter decimal number: "))
decimal_binary(n)
```

```python
#Find Average of List
def average(lst):
    print(sum(lst) / len(lst))

l = list(map(int, input("Enter numbers: ").split()))
average(l)
```

```python
#Check Number Ends With 5
def ends_with_5(num):
    if str(num).endswith("5"):
        print("Yes")
    else:
        print("No")

n = int(input("Enter number: "))
ends_with_5(n)
```

```python
#Print ASCII Values
def ascii_values(text):
    for ch in text:
        print(ch, "=", ord(ch))

s = input("Enter text: ")
ascii_values(s)
```

```python
#Find Length of List
def list_length(lst):
    print(len(lst))

l = list(map(int, input("Enter numbers: ").split()))
list_length(l)
```

```python
#Find Largest Word
def largest_word(text):
    words = text.split()
    print(max(words, key=len))

s = input("Enter sentence: ")
largest_word(s)
```
