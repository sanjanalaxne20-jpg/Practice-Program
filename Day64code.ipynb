```python
#Shaft Power Calculation
torque = 250
rpm = 1500

power = (2 * 3.14159 * rpm * torque) / 60

print("Shaft Power:", round(power / 1000, 2), "kW")
```

```python
#Stress and Strain
force = 20000
area = 400
original_length = 2
extension = 0.004

stress = force / area
strain = extension / original_length

print("Stress:", stress, "MPa")
print("Strain:", strain)
```

```python
#Young's Modulus
stress = 50
strain = 0.002

youngs_modulus = stress / strain

print("Young's Modulus:", youngs_modulus, "MPa")
```

```python
#Engine Thermal Efficiency
input_energy = 5000
useful_energy = 1800

efficiency = (useful_energy / input_energy) * 100
heat_loss = input_energy - useful_energy

print("Thermal Efficiency:", efficiency, "%")
print("Energy Loss:", heat_loss, "kJ")
```

```python
#Pipe Flow Velocity
flow_rate = 0.05
area = 0.01

velocity = flow_rate / area

print("Flow Velocity:", velocity, "m/s")
```

```python
#Reynolds Number
density = 1000
velocity = 2
diameter = 0.05
viscosity = 0.001

reynolds = (density * velocity * diameter) / viscosity

print("Reynolds Number:", reynolds)

if reynolds < 2300:
    print("Flow: Laminar")
elif reynolds < 4000:
    print("Flow: Transitional")
else:
    print("Flow: Turbulent")
```

```python
#Voltage Drop in Cable
current = 20
resistance = 0.08

voltage_drop = current * resistance

print("Voltage Drop:", voltage_drop, "V")
```

```python
#Power Loss in Transmission
current = 100
resistance = 0.5

power_loss = current ** 2 * resistance

print("Transmission Power Loss:", power_loss, "W")
```

```python
#Solar Energy Savings
solar_generation = 40
electricity_rate = 8
days = 30

monthly_savings = solar_generation * electricity_rate * days

print("Monthly Energy Savings: ₹", monthly_savings)
```

```python
#Production Break-Even Analysis
fixed_cost = 200000
selling_price = 500
variable_cost = 300

break_even_units = fixed_cost / (selling_price - variable_cost)

print("Break-Even Units:", int(break_even_units))
```
