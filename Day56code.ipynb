```python
#Machine Efficiency
input_power = 5000
output_power = 4250
efficiency = (output_power / input_power) * 100
power_loss = input_power - output_power
print("Machine Efficiency:", efficiency, "%")
print("Power Loss:", power_loss, "W")
```

```python
#Battery Backup Calculation
battery_capacity = 5000
device_power = 250
efficiency = 0.85
backup_hours = (battery_capacity * efficiency) / device_power
print("Estimated Battery Backup:", round(backup_hours, 2), "hours")
```

```python
#Concrete Requirement
length = 12
width = 5
depth = 0.15
volume = length * width * depth
concrete_needed = volume * 2400
print("Concrete Volume:", volume, "m³")
print("Concrete Weight:", concrete_needed, "kg")
```

```python
#Solar Panel Power Generation
panels = 20
panel_power = 400
sun_hours = 5
efficiency = 0.82
daily_energy = panels * panel_power * sun_hours * efficiency
print("Daily Energy Generated:", daily_energy / 1000, "kWh")
```

```python
#Vehicle Speed Analysis
distance = 240
time = 4
speed = distance / time
if speed > 60:
    status = "High Speed"
elif speed >= 40:
    status = "Normal Speed"
else:
    status = "Low Speed"
print("Average Speed:", speed, "km/h")
print("Status:", status)
```

```python
#Production Capacity
machines = 8
units_per_hour = 45
working_hours = 10
efficiency = 0.90
production = machines * units_per_hour * working_hours * efficiency
print("Expected Production:", int(production), "units")
```

```python
#Temperature Monitoring System
temperatures = [72, 75, 81, 85, 79, 88, 76]
limit = 80
for i, temp in enumerate(temperatures, start=1):
    if temp > limit:
        print("Hour", i, ":", temp, "°C - WARNING")
    else:
        print("Hour", i, ":", temp, "°C - Normal")
```

```python
#Predictive Maintenance
machines = {
    "Machine A": {"hours": 4200, "vibration": 3.2},
    "Machine B": {"hours": 1800, "vibration": 1.5},
    "Machine C": {"hours": 3900, "vibration": 4.1},
    "Machine D": {"hours": 2500, "vibration": 2.0}
}

for machine, data in machines.items():

    if data["hours"] > 4000 or data["vibration"] > 3:
        status = "Maintenance Required"
    else:
        status = "Operating Normally"
    print(machine, "->", status)
```

```python
#Engineering Project Cost Estimation
materials = {
    "Steel": 85000,
    "Cement": 45000,
    "Electrical": 30000,
    "Labor": 65000
}
material_cost = sum(materials.values())
contingency = material_cost * 0.10
total_cost = material_cost + contingency
print("Base Project Cost:", material_cost)
print("Contingency:", contingency)
print("Total Project Cost:", total_cost)
```
