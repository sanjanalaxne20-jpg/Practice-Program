```python
#Filter Vowels
text = input("Enter characters: ").split()
result = list(filter(lambda x: x.lower() in "aeiou", text))
print(result)
```

```python
#Filter Consonants
text = input("Enter characters: ").split()
result = list(filter(lambda x: x.lower() not in "aeiou", text))
print(result)
```

```python
#Filter Non-Zero Numbers
l1 = list(map(int, input("Enter numbers: ").split()))
result = list(filter(lambda x: x != 0, l1))
print(result)
```

```python
#Filter Numbers Divisible by 7
l1 = list(map(int, input("Enter numbers: ").split()))
result = list(filter(lambda x: x % 7 == 0, l1))
print(result)
```

```python
#Filter Numbers Between 10 and 50
l1 = list(map(int, input("Enter numbers: ").split()))
result = list(filter(lambda x: 10 <= x <= 50, l1))
print(result)
```

```python
#Filter Uppercase Words
words = input("Enter words: ").split()
result = list(filter(lambda x: x.isupper(), words))
print(result)
```

```python
#Filter Lowercase Words
words = input("Enter words: ").split()
result = list(filter(lambda x: x.islower(), words))
print(result)
```

```python
#Filter Numbers Ending with 5
l1 = list(map(int, input("Enter numbers: ").split()))
result = list(filter(lambda x: str(x).endswith("5"), l1))
print(result)
```

```python
#Filter Names Containing Letter 'a'
names = input("Enter names: ").split()
result = list(filter(lambda x: 'a' in x.lower(), names))
print(result)
```

```python
#Filter Empty Strings
words = ["Python", "", "Java", "", "SQL"]
result = list(filter(None, words))
print(result)
```
