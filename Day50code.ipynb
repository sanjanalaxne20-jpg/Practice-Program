```python
#Employee Commission
sales = float(input("Enter total sales: "))

if sales >= 100000:
    commission = sales * 0.10
elif sales >= 50000:
    commission = sales * 0.07
else:
    commission = sales * 0.03

print("Commission = ₹", commission)
```

```python
#Mobile Storage Alert
total = float(input("Enter total storage in GB: "))
used = float(input("Enter used storage in GB: "))

usage = (used / total) * 100

print("Storage Used =", round(usage, 2), "%")

if usage >= 90:
    print("Storage Almost Full")
else:
    print("Storage Available")
```

```python
#Train Platform Allocation
train_number = int(input("Enter train number: "))
platform = int(input("Enter platform number: "))

print("Train", train_number, "is assigned to Platform", platform)
```

```python
#Sales Target Achievement
target = float(input("Enter sales target: "))
sales = float(input("Enter actual sales: "))

achievement = (sales / target) * 100

print("Target Achievement =", round(achievement, 2), "%")

if sales >= target:
    print("Target Achieved")
else:
    print("Target Not Achieved")
```

```python
#Website Conversion Rate
visitors = int(input("Enter website visitors: "))
customers = int(input("Enter customers: "))

conversion = (customers / visitors) * 100

print("Conversion Rate =", round(conversion, 2), "%")
```

```python
#Product Return Rate
sold = int(input("Enter products sold: "))
returned = int(input("Enter products returned: "))

return_rate = (returned / sold) * 100

print("Return Rate =", round(return_rate, 2), "%")
```

```python
#Employee Productivity
hours = float(input("Enter working hours: "))
tasks = int(input("Enter tasks completed: "))

productivity = tasks / hours

print("Productivity =", round(productivity, 2), "tasks/hour")
```

```python
#Cricket Run Rate
runs = int(input("Enter total runs: "))
overs = float(input("Enter overs played: "))

run_rate = runs / overs

print("Run Rate =", round(run_rate, 2))
```

```python
#Data Usage Percentage
total_data = float(input("Enter total data (GB): "))
used_data = float(input("Enter used data (GB): "))

percentage = (used_data / total_data) * 100

print("Data Used =", round(percentage, 2), "%")
```

```python
#Battery Health
original = int(input("Enter original battery capacity: "))
current = int(input("Enter current battery capacity: "))

health = (current / original) * 100

print("Battery Health =", round(health, 2), "%")
```
