```python
#Clothing Store Bill
shirts = int(input("Number of shirts: "))
price = 700

bill = shirts * price

print("Bill Amount = ₹", bill)
```

```python
#Online Exam Result
correct = int(input("Correct answers: "))
wrong = int(input("Wrong answers: "))

score = (correct * 2) - wrong

print("Final Score =", score)
```

```python
#Factory Machine Production
machines = int(input("Number of machines: "))
items = int(input("Items produced per machine: "))

production = machines * items

print("Total Production =", production)
```

```python
#Daily Steps Tracker
steps = int(input("Enter today's steps: "))

if steps >= 10000:
    print("Daily Goal Achieved!")
else:
    print("Keep Walking!")
```

```python
#Bank Deposit Calculator
deposit = float(input("Enter deposit amount: "))
interest = deposit * 0.05

print("Interest =", interest)
print("Total Amount =", deposit + interest)
```

```python
#Cinema Ticket Revenue
tickets = int(input("Enter tickets sold: "))
price = float(input("Enter ticket price: "))

revenue = tickets * price

print("Total Revenue =", revenue)
```

```python
#Mobile Shop Profit
cost = float(input("Enter cost price: "))
sell = float(input("Enter selling price: "))

profit = sell - cost

print("Profit =", profit)
```

```python
#Office Stationery Cost
pens = int(input("Enter number of pens: "))
notebooks = int(input("Enter number of notebooks: "))

bill = (pens * 10) + (notebooks * 50)

print("Total Bill =", bill)
```

```python
#Water Tank Filling Time
capacity = float(input("Tank capacity (liters): "))
speed = float(input("Water flow per minute: "))

time = capacity / speed

print("Time Required =", time, "minutes")
```
