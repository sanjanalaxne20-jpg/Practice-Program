```python
#Production Defect Rate
total_products = 10000
defective_products = 240
defect_rate = (defective_products / total_products) * 100
good_products = total_products - defective_products
print("Defect Rate:", defect_rate, "%")
print("Good Products:", good_products)
```

```python
#Vehicle Acceleration
initial_speed = 10
final_speed = 30
time = 5
acceleration = (final_speed - initial_speed) / time
print("Acceleration:", acceleration, "m/s²")
```

```python
#Road Gradient
height = 12
horizontal_distance = 300
gradient = (height / horizontal_distance) * 100
print("Road Gradient:", round(gradient, 2), "%")
```

```python
#Elevator Power Requirement
mass = 800
height = 20
time = 25
gravity = 9.81
power = (mass * gravity * height) / time
print("Required Power:", round(power / 1000, 2), "kW")
```

```python
#Temperature Conversion Analysis
celsius = 75
fahrenheit = (celsius * 9 / 5) + 32
kelvin = celsius + 273.15
print("Celsius:", celsius)
print("Fahrenheit:", fahrenheit)
print("Kelvin:", kelvin)
```

```python
#Material Safety Factor
maximum_strength = 500
working_stress = 125
safety_factor = maximum_strength / working_stress
print("Safety Factor:", safety_factor)
if safety_factor >= 4:
    print("Design is Safe")
else:
    print("Design Needs Review")
```

```python
#Water Pump Efficiency
input_power = 5000
useful_output = 4000
efficiency = (useful_output / input_power) * 100
print("Pump Efficiency:", efficiency, "%")
```

```python
#Conveyor Belt Capacity
items_per_minute = 35
working_hours = 8
total_items = items_per_minute * 60 * working_hours
print("Daily Conveyor Capacity:", total_items, "items")
```

```python
#Average Material Strength
strength_values = [420, 435, 428, 440, 432]
average_strength = sum(strength_values) / len(strength_values)
print("Average Material Strength:",
      round(average_strength, 2), "MPa")
```

```python
#Engineering Cost Comparison
projects = {
    "Project A": 450000,
    "Project B": 520000,
    "Project C": 390000,
    "Project D": 610000
}

cheapest = min(projects, key=projects.get)
most_expensive = max(projects, key=projects.get)
print("Cheapest Project:", cheapest)
print("Cost: ₹", projects[cheapest])
print("Most Expensive Project:", most_expensive)
print("Cost: ₹", projects[most_expensive])
```
