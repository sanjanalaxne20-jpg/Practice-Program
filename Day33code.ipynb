```python
#Filter Even Numbers
l1 = list(map(int, input("Enter numbers: ").split()))
result = list(filter(lambda x: x % 2 == 0, l1))
print(result)
```

```python
#Filter Odd Numbers
l1 = list(map(int, input("Enter numbers: ").split()))
result = list(filter(lambda x: x % 2 != 0, l1))
print(result)
```

```python
#Filter Positive Numbers
l1 = list(map(int, input("Enter numbers: ").split()))
result = list(filter(lambda x: x > 0, l1))
print(result)
```

```python
#Filter Negative Numbers
l1 = list(map(int, input("Enter numbers: ").split()))
result = list(filter(lambda x: x < 0, l1))
print(result)
```

```python
#Filter Multiples of 5
l1 = list(map(int, input("Enter numbers: ").split()))
result = list(filter(lambda x: x % 5 == 0, l1))
print(result)
```

```python
#Filter Strings Longer Than 5 Characters
words = input("Enter words: ").split()
result = list(filter(lambda x: len(x) > 5, words))
print(result)
```

```python
#Filter Words Starting with 'A'
words = input("Enter words: ").split()
result = list(filter(lambda x: x.lower().startswith('a'), words))
print(result)
```
