```python
#Bolt Strength Analysis
force = 12000
area = 150
stress = force / area
print("Applied Force:", force, "N")
print("Bolt Area:", area, "mm²")
print("Stress:", round(stress, 2), "N/mm²")
if stress > 80:
    print("Warning: High Stress")
else:
    print("Bolt is Safe")
```

```python
#Building Floor Area
length = 30
width = 20
floors = 5
floor_area = length * width
total_area = floor_area * floors
print("Area of One Floor:", floor_area, "m²")
print("Total Building Area:", total_area, "m²")
```

```python
#Water Tank Capacity
length = 4
width = 3
height = 2.5
volume = length * width * height
liters = volume * 1000
print("Tank Volume:", volume, "m³")
print("Capacity:", liters, "liters")
```

```python
#Gear Ratio
driver_teeth = 20
driven_teeth = 60
driver_speed = 1500
gear_ratio = driven_teeth / driver_teeth
output_speed = driver_speed / gear_ratio
print("Gear Ratio:", gear_ratio)
print("Output Speed:", output_speed, "RPM")
```

```python
#Electrical Resistance
voltage = 230
current = 5
resistance = voltage / current
print("Voltage:", voltage, "V")
print("Current:", current, "A")
print("Resistance:", resistance, "Ohms")
```

```python
#Electrical Power
voltage = 240
current = 8
power_factor = 0.9
power = voltage * current * power_factor
print("Power Consumption:", power, "W")
print("Power Consumption:", power / 1000, "kW")
```

```python
#Train Travel Analysis
distance = 450
time = 5.5
speed = distance / time
print("Distance:", distance, "km")
print("Travel Time:", time, "hours")
print("Average Speed:", round(speed, 2), "km/h")
```

```python
#Pump Flow Rate
water_volume = 18000
time = 30
flow_rate = water_volume / time
print("Water Volume:", water_volume, "liters")
print("Time:", time, "minutes")
print("Flow Rate:", flow_rate, "liters/minute")
```

```python
#Machine Downtime Analysis
total_hours = 720
downtime = 36
availability = ((total_hours - downtime) / total_hours) * 100
print("Total Hours:", total_hours)
print("Downtime:", downtime)
print("Machine Availability:", round(availability, 2), "%")
```

```python
#Factory OEE Calculation
availability = 95
performance = 90
quality = 98
oee = (availability * performance * quality) / 10000
print("Availability:", availability, "%")
print("Performance:", performance, "%")
print("Quality:", quality, "%")
print("OEE:", round(oee, 2), "%")
```
