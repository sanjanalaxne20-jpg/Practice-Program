```python
#Petrol Pump
amount = float(input("Enter amount: "))
price = float(input("Price per liter: "))

liters = amount / price

print("Petrol Received =", liters, "liters")
```

```python
#Warehouse Boxes
boxes = int(input("Number of boxes: "))
weight = float(input("Weight per box: "))

total_weight = boxes * weight

print("Total Weight =", total_weight)
```

```python
#Flight Ticket Discount
ticket = float(input("Ticket Price: "))
discount = float(input("Discount (%): "))

final = ticket - (ticket * discount / 100)

print("Final Ticket Price =", final)
```

```python
#Farmer Profit
income = float(input("Income: "))
expense = float(input("Expense: "))

profit = income - expense

print("Profit =", profit)
```

```python
#Electricity Power Consumption
power = float(input("Power Rating (kW): "))
hours = float(input("Hours Used: "))

units = power * hours

print("Electricity Consumed =", units, "kWh")
```

```python
#Cinema Seat Booking
total_seats = int(input("Enter total seats: "))
booked = int(input("Enter booked seats: "))

available = total_seats - booked

print("Available Seats =", available)
```

```python
#Mobile Battery Percentage
battery = int(input("Enter battery percentage: "))

if battery < 20:
    print("Charge your phone")
else:
    print("Battery is sufficient")
```
