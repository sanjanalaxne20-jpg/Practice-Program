```python
#Machine Utilization Rate
machines = {
    "CNC-01": [20, 18],
    "CNC-02": [20, 15],
    "CNC-03": [20, 19],
    "CNC-04": [20, 16]
}
for machine, values in machines.items():
    available = values[0]
    running = values[1]

    utilization = (running / available) * 100

    print(machine, ":", round(utilization, 2), "%")
```

```python
#Bearing Life Estimation
dynamic_load = 10000
actual_load = 2500
base_life = 1000

life = base_life * (dynamic_load / actual_load) ** 3

print("Estimated Bearing Life:",
      round(life, 2), "hours")
```

```python
#Energy Cost Optimization
machines = {
    "Motor": {"power": 10, "hours": 8},
    "Pump": {"power": 6, "hours": 5},
    "Compressor": {"power": 15, "hours": 4}
}

electricity_rate = 8
total_cost = 0

for machine, data in machines.items():
    energy = data["power"] * data["hours"]
    cost = energy * electricity_rate
    total_cost += cost

    print(machine, "Cost: ₹", cost)

print("Total Daily Cost: ₹", total_cost)
```

```python
#Maintenance Priority System
machines = {
    "Machine A": {"age": 8, "failures": 2},
    "Machine B": {"age": 4, "failures": 0},
    "Machine C": {"age": 10, "failures": 5},
    "Machine D": {"age": 6, "failures": 1}
}

for machine, data in machines.items():

    score = data["age"] + data["failures"] * 2

    if score >= 15:
        priority = "HIGH"
    elif score >= 8:
        priority = "MEDIUM"
    else:
        priority = "LOW"

    print(machine, "->", priority,
          "| Score:", score)
```

```python
#IoT Sensor Anomaly Detection
readings = [20, 21, 22, 23, 75, 21, 22, 80, 23]

average = sum(readings) / len(readings)

print("Average:", round(average, 2))

for value in readings:
    if abs(value - average) > 30:
        print("Anomaly Detected:", value)
```

```python
#Beam Load Analysis
loads = [1200, 1500, 1800, 1350, 1600]

maximum_load = max(loads)
average_load = sum(loads) / len(loads)

print("Maximum Load:", maximum_load, "N")
print("Average Load:", round(average_load, 2), "N")

if maximum_load > 1700:
    print("Warning: High Load")
else:
    print("Load is Safe")
```

```python
#Battery Degradation Analysis
initial_capacity = 100
yearly_capacity = [100, 96, 91, 87, 82, 77]

for year, capacity in enumerate(yearly_capacity):
    degradation = initial_capacity - capacity

    print("Year", year,
          "| Capacity:", capacity, "%",
          "| Degradation:", degradation, "%")
```

```python
#Vehicle Maintenance Prediction
vehicles = {
    "Car-01": {"km": 45000, "age": 3},
    "Car-02": {"km": 85000, "age": 6},
    "Car-03": {"km": 30000, "age": 2}
}

for vehicle, data in vehicles.items():

    if data["km"] > 80000 or data["age"] > 5:
        status = "Service Required"
    else:
        status = "No Major Service Required"

    print(vehicle, "->", status)
```

```python
#Material Quality Classification
materials = {
    "Steel A": 450,
    "Steel B": 520,
    "Steel C": 390,
    "Steel D": 480
}

required_strength = 400

for material, strength in materials.items():

    if strength >= required_strength:
        result = "PASS"
    else:
        result = "FAIL"

    print(material,
          "| Strength:", strength,
          "MPa |", result)
```
