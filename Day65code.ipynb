```python
#Robot Battery Scheduling
robots = {
    "Robot-A": 85,
    "Robot-B": 42,
    "Robot-C": 68,
    "Robot-D": 25
}

for robot, battery in robots.items():

    if battery < 30:
        action = "Charge Now"
    elif battery < 50:
        action = "Schedule Charging"
    else:
        action = "Continue Work"

    print(robot, ":", battery, "% ->", action)
```

```python
#IoT Sensor Health Score
sensors = {
    "Sensor-A": {"battery": 90, "accuracy": 96},
    "Sensor-B": {"battery": 45, "accuracy": 82},
    "Sensor-C": {"battery": 75, "accuracy": 94}
}

for sensor, data in sensors.items():

    score = (data["battery"] + data["accuracy"]) / 2

    if score >= 90:
        status = "Excellent"
    elif score >= 75:
        status = "Good"
    else:
        status = "Needs Maintenance"

    print(sensor, "Score:", round(score, 2), "->", status)
```

```python
#Electrical Load Balancing
loads = {
    "Phase A": 42,
    "Phase B": 38,
    "Phase C": 65
}

average = sum(loads.values()) / len(loads)

print("Average Load:", round(average, 2), "A")

for phase, load in loads.items():

    difference = load - average

    print(
        phase,
        "| Load:", load,
        "A | Difference:", round(difference, 2), "A"
    )
```

```python
#Production Target Calculator
daily_target = 5000
working_days = 26
current_production = 118000

monthly_target = daily_target * working_days
remaining = monthly_target - current_production

print("Monthly Target:", monthly_target)
print("Current Production:", current_production)
print("Remaining Production:", remaining)

if remaining <= 0:
    print("Target Achieved")
else:
    print("Target Not Achieved")
```

```python
#Tool Wear Monitoring
tools = {
    "Tool-1": 12,
    "Tool-2": 28,
    "Tool-3": 45,
    "Tool-4": 18
}

maximum_wear = 30

for tool, wear in tools.items():

    if wear > maximum_wear:
        status = "Replace"
    elif wear > 20:
        status = "Monitor"
    else:
        status = "Good"

    print(tool, "Wear:", wear, "% ->", status)
```

```python
#Concrete Strength Test
samples = [28, 31, 29, 32, 27, 30]

required_strength = 30
average_strength = sum(samples) / len(samples)

print("Average Strength:",
      round(average_strength, 2), "MPa")

if average_strength >= required_strength:
    print("Concrete: PASS")
else:
    print("Concrete: FAIL")
```

```python
#Vehicle Battery Range
battery_capacity = 60
consumption = 0.15

range_km = battery_capacity / consumption

print("Estimated Vehicle Range:",
      round(range_km, 2), "km")
```

```python
#Charging Cost Calculator
battery_capacity = 60
electricity_rate = 9
charging_efficiency = 0.9

energy_required = battery_capacity / charging_efficiency
cost = energy_required * electricity_rate

print("Energy Required:",
      round(energy_required, 2), "kWh")
print("Charging Cost: ₹", round(cost, 2))
```

```python
#Pump Energy Consumption
power = 7.5
hours_per_day = 6
days = 30
rate = 8

energy = power * hours_per_day * days
cost = energy * rate

print("Monthly Energy:", energy, "kWh")
print("Monthly Cost: ₹", cost)
```

```python
#Traffic Average Speed
vehicles = {
    "Car": {"distance": 50, "time": 1},
    "Bus": {"distance": 40, "time": 1.2},
    "Truck": {"distance": 60, "time": 1.5}
}

for vehicle, data in vehicles.items():

    speed = data["distance"] / data["time"]

    print(
        vehicle,
        "Average Speed:",
        round(speed, 2),
        "km/h"
    )
```
