```python
#HVAC Cooling Load
room_area = 500
people = 30
equipment_load = 5000

area_load = room_area * 10
people_load = people * 100

total_load = area_load + people_load + equipment_load

print("Cooling Load:", total_load, "W")
print("Cooling Load:", total_load / 1000, "kW")
```

```python
#Motor RPM Analysis
motor_speeds = [1450, 1480, 1465, 1420, 1490]

average = sum(motor_speeds) / len(motor_speeds)
minimum = min(motor_speeds)
maximum = max(motor_speeds)

print("Average RPM:", round(average, 2))
print("Minimum RPM:", minimum)
print("Maximum RPM:", maximum)
```

```python
#Fuel Consumption Analysis
fuel_data = {
    "Generator 1": 12,
    "Generator 2": 15,
    "Generator 3": 10,
    "Generator 4": 18
}

total = sum(fuel_data.values())
average = total / len(fuel_data)

print("Total Fuel:", total, "liters")
print("Average Fuel:", average, "liters")

highest = max(fuel_data, key=fuel_data.get)

print("Highest Consumption:", highest)
```

```python
#Construction Progress
construction = {
    "Foundation": 100,
    "Columns": 85,
    "Walls": 70,
    "Roof": 45,
    "Electrical": 30
}

progress = sum(construction.values()) / len(construction)

print("Overall Construction Progress:",
      round(progress, 2), "%")
```

```python
#Network Bandwidth Utilization
bandwidth = 1000
traffic = [650, 720, 850, 950, 780]

for value in traffic:
    utilization = (value / bandwidth) * 100

    print("Traffic:", value,
          "Mbps | Utilization:",
          utilization, "%")

    if utilization > 90:
        print("WARNING: Network Overloaded")
```

```python
#Server Load Monitoring
servers = {
    "Server-1": 45,
    "Server-2": 72,
    "Server-3": 91,
    "Server-4": 60
}

for server, load in servers.items():

    if load >= 90:
        status = "Critical"
    elif load >= 70:
        status = "High"
    else:
        status = "Normal"

    print(server, ":", load, "% ->", status)
```

```python
#Engineering Trend Analysis
production = [1200, 1350, 1400, 1550, 1700, 1850]

for i in range(1, len(production)):
    change = production[i] - production[i - 1]

    print(
        "Period", i + 1,
        "Change:", change,
        "units"
    )
```

```python
#Traffic Density Analysis
traffic = {
    "8 AM": 850,
    "10 AM": 520,
    "1 PM": 680,
    "5 PM": 1100,
    "8 PM": 900
}

peak_time = max(traffic, key=traffic.get)

print("Peak Traffic Time:", peak_time)
print("Vehicles:", traffic[peak_time])
```

```python
#Water Quality Monitoring
water_samples = {
    "Sample A": {"ph": 7.1, "turbidity": 2},
    "Sample B": {"ph": 5.8, "turbidity": 8},
    "Sample C": {"ph": 7.4, "turbidity": 3}
}

for sample, data in water_samples.items():

    if 6.5 <= data["ph"] <= 8.5 and data["turbidity"] <= 5:
        status = "Safe"
    else:
        status = "Needs Treatment"

    print(sample, "->", status)
```

```python
#Robot Battery & Operation Analysis
robots = {
    "Robot A": {"battery": 90, "tasks": 45},
    "Robot B": {"battery": 55, "tasks": 30},
    "Robot C": {"battery": 25, "tasks": 18}
}

for robot, data in robots.items():

    if data["battery"] < 30:
        action = "Recharge Immediately"
    elif data["battery"] < 60:
        action = "Recharge Soon"
    else:
        action = "Continue Operation"

    print(robot)
    print("Battery:", data["battery"], "%")
    print("Completed Tasks:", data["tasks"])
    print("Action:", action)
    print("--------------------")
```
