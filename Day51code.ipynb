```python
#Employee Age for Retirement
age = int(input("Enter employee age: "))
retirement = 60

remaining = retirement - age

if remaining > 0:
    print("Years Until Retirement =", remaining)
else:
    print("Retirement Age Reached")
```

```python
#Train Journey Distance
speed = float(input("Enter train speed (km/h): "))
time = float(input("Enter journey time (hours): "))

distance = speed * time

print("Distance Covered =", distance, "km")
```

```python
#Social Media Engagement
likes = int(input("Enter likes: "))
comments = int(input("Enter comments: "))
shares = int(input("Enter shares: "))

engagement = likes + comments + shares

print("Total Engagement =", engagement)
```

```python
#Employee Attrition Rate
employees = int(input("Enter total employees: "))
left = int(input("Enter employees who left: "))

attrition = (left / employees) * 100

print("Attrition Rate =", round(attrition, 2), "%")
```

```python
#Crop Yield per Acre
production = float(input("Enter total crop production (kg): "))
acres = float(input("Enter land area (acres): "))

yield_per_acre = production / acres

print("Crop Yield =", yield_per_acre, "kg/acre")
```

```python
#Hospital Bed Occupancy
beds = int(input("Enter total beds: "))
occupied = int(input("Enter occupied beds: "))

occupancy = (occupied / beds) * 100

print("Bed Occupancy =", round(occupancy, 2), "%")
```

```python
#Delivery Success Rate
total = int(input("Enter total deliveries: "))
successful = int(input("Enter successful deliveries: "))

rate = (successful / total) * 100

print("Delivery Success Rate =", round(rate, 2), "%")
```

```python
#Student Improvement
old_marks = float(input("Enter previous marks: "))
new_marks = float(input("Enter current marks: "))

improvement = new_marks - old_marks

print("Marks Improvement =", improvement)

if improvement > 0:
    print("Performance Improved")
elif improvement < 0:
    print("Performance Decreased")
else:
    print("No Change")
```

```python
#Product Profit Margin
cost = float(input("Enter product cost: "))
selling = float(input("Enter selling price: "))

profit = selling - cost
margin = (profit / selling) * 100

print("Profit = ₹", profit)
print("Profit Margin =", round(margin, 2), "%")
```

```python
#Monthly Savings
income = float(input("Enter monthly income: "))
expenses = float(input("Enter monthly expenses: "))

savings = income - expenses
saving_rate = (savings / income) * 100

print("Monthly Savings = ₹", savings)
print("Saving Rate =", round(saving_rate, 2), "%")
```
