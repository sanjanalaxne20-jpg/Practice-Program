```python
#Torque Calculation
force = 250
radius = 0.4
torque = force * radius
print("Force:", force, "N")
print("Radius:", radius, "m")
print("Torque:", torque, "N-m")
```

```python
#Mechanical Work
force = 500
distance = 12
work = force * distance
print("Mechanical Work:", work, "J")
```

```python
#Kinetic Energy
mass = 1200
velocity = 20
kinetic_energy = 0.5 * mass * velocity ** 2
print("Kinetic Energy:", kinetic_energy, "J")
```

```python
#Potential Energy
mass = 500
height = 15
gravity = 9.81
potential_energy = mass * gravity * height
print("Potential Energy:", round(potential_energy, 2), "J")
```

```python
#Spring Force
spring_constant = 250
extension = 0.08
force = spring_constant * extension
print("Spring Force:", force, "N")
```

```python
#Pipe Pressure
force = 15000
area = 0.5
pressure = force / area
print("Pressure:", pressure, "Pa")
```

```python
#Transformer Voltage
primary_voltage = 230
primary_turns = 1000
secondary_turns = 200
secondary_voltage = (
    primary_voltage * secondary_turns / primary_turns
)
print("Secondary Voltage:", secondary_voltage, "V")
```

```python
#Three-Phase Power
voltage = 415
current = 20
power_factor = 0.85
power = (3 ** 0.5) * voltage * current * power_factor
print("Three-Phase Power:", round(power / 1000, 2), "kW")
```

```python
#Battery State of Charge
capacity = 100
remaining = 72
soc = (remaining / capacity) * 100
print("Battery State of Charge:", soc, "%")
if soc >= 80:
    print("Battery Level: High")
elif soc >= 40:
    print("Battery Level: Medium")
else:
    print("Battery Level: Low")
```

```python
#Solar Panel Efficiency
solar_input = 1000
electrical_output = 180
efficiency = (electrical_output / solar_input) * 100
print("Solar Panel Efficiency:", efficiency, "%")
```
