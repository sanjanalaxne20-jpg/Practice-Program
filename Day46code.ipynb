```python
#Bike Rental Charge
hours = int(input("Enter rental hours: "))
rate = 100

amount = hours * rate

print("Rental Charge = ₹", amount)
```

```python
#School Fee Balance
total_fee = float(input("Enter total fee: "))
paid = float(input("Enter paid amount: "))

balance = total_fee - paid

print("Fee Balance = ₹", balance)
```

```python
#Mobile Shop Discount
amount = float(input("Enter purchase amount: "))

if amount >= 20000:
    discount = amount * 0.15
else:
    discount = amount * 0.05

print("Discount = ₹", discount)
print("Final Amount = ₹", amount - discount)
```

```python
#Gas Cylinder Cost
cylinders = int(input("Enter number of cylinders: "))
price = 1100

total = cylinders * price

print("Total Cost = ₹", total)
```

```python
#Water Park Ticket
adults = int(input("Enter number of adults: "))
children = int(input("Enter number of children: "))

bill = (adults * 600) + (children * 350)

print("Total Ticket Cost = ₹", bill)
```

```python
#Exam Answer Checker
correct = int(input("Enter correct answers: "))
total = int(input("Enter total questions: "))

percentage = (correct / total) * 100

print("Score Percentage =", percentage)
```

```python
#Toll Tax Calculator
vehicle = input("Enter vehicle type (car/bus/truck): ")

if vehicle == "car":
    print("Toll Tax = ₹100")
elif vehicle == "bus":
    print("Toll Tax = ₹200")
else:
    print("Toll Tax = ₹300")
```

```python
#Pizza Order Bill
pizza = int(input("Enter number of pizzas: "))
price = 299

bill = pizza * price

print("Total Bill = ₹", bill)
```

```python
#Blood Donation Eligibility
age = int(input("Enter age: "))
weight = float(input("Enter weight: "))

if age >= 18 and weight >= 50:
    print("Eligible to Donate Blood")
else:
    print("Not Eligible")
```
