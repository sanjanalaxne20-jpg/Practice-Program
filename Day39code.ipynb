```python
#Water Bill
liters = int(input("Enter water consumed: "))
rate = float(input("Rate per liter: "))

bill = liters * rate

print("Water Bill =", bill)
```

```python
#Employee Overtime
hours = int(input("Overtime hours: "))
rate = float(input("Overtime rate: "))

pay = hours * rate

print("Overtime Pay =", pay)
```

```python
#Courier Charge
weight = float(input("Parcel weight (kg): "))
rate = float(input("Charge per kg: "))

charge = weight * rate

print("Courier Charge =", charge)
```

```python
#Internet Café Bill
hours = int(input("Hours used: "))
rate = float(input("Charge per hour: "))

bill = hours * rate

print("Total Bill =", bill)
```

```python
#Cricket Strike Rate
runs = int(input("Runs scored: "))
balls = int(input("Balls faced: "))

strike_rate = (runs / balls) * 100

print("Strike Rate =", strike_rate)
```

```python
#Exam Result
total = int(input("Total Marks: "))
obtained = int(input("Marks Obtained: "))

percentage = (obtained / total) * 100

print("Percentage =", percentage)
```

```python
#Loan Interest
loan = float(input("Loan Amount: "))
rate = float(input("Interest Rate: "))

interest = loan * rate / 100

print("Interest =", interest)
```

```python
#Taxi Fare
distance = float(input("Distance (km): "))
fare = float(input("Fare per km: "))

total = distance * fare

print("Taxi Fare =", total)
```

```python
#Factory Production
workers = int(input("Number of workers: "))
items = int(input("Items produced per worker: "))

total = workers * items

print("Total Production =", total)
```

```python
#Online Course Fee
courses = int(input("Number of courses: "))
fee = float(input("Fee per course: "))

total = courses * fee

print("Total Fee =", total)
```
