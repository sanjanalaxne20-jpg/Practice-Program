```python
#Machine Health Monitoring System
machines = {
    "Machine A": {"temperature": 72, "vibration": 2.1, "hours": 2500},
    "Machine B": {"temperature": 91, "vibration": 4.5, "hours": 4200},
    "Machine C": {"temperature": 68, "vibration": 1.8, "hours": 1800},
    "Machine D": {"temperature": 85, "vibration": 3.7, "hours": 3900}
}

for machine, data in machines.items():
    temperature = data["temperature"]
    vibration = data["vibration"]
    hours = data["hours"]

    if temperature > 85 or vibration > 3.5 or hours > 4000:
        status = "Maintenance Required"
    else:
        status = "Machine Healthy"
    print(machine)
    print("Temperature:", temperature, "°C")
    print("Vibration:", vibration, "mm/s")
    print("Operating Hours:", hours)
    print("Status:", status)
    print("-------------------------")
```

```python
#Electricity Consumption Analyzer
monthly_units = {
    "January": 420,
    "February": 390,
    "March": 450,
    "April": 520,
    "May": 610,
    "June": 580
}
rate = 8
total_units = sum(monthly_units.values())
total_bill = total_units * rate
average_units = total_units / len(monthly_units)

highest_month = max(monthly_units, key=monthly_units.get)
lowest_month = min(monthly_units, key=monthly_units.get)

print("Total Electricity Consumption:", total_units, "units")
print("Total Electricity Bill: ₹", total_bill)
print("Average Monthly Consumption:", round(average_units, 2), "units")

print("Highest Consumption:", highest_month)
print("Units:", monthly_units[highest_month])

print("Lowest Consumption:", lowest_month)
print("Units:", monthly_units[lowest_month])
```

```python
#Solar Power Generation Analyzer
panels = 25
panel_capacity = 400
sun_hours = 5.5
efficiency = 0.82


daily_generation = (
    panels *
    panel_capacity *
    sun_hours *
    efficiency
) / 1000


monthly_generation = daily_generation * 30
print("Number of Panels:", panels)
print("Panel Capacity:", panel_capacity, "W")
print("Daily Generation:", round(daily_generation, 2), "kWh")
print("Monthly Generation:", round(monthly_generation, 2), "kWh")
```

```python
#Production Efficiency Analyzer
machines = {
    "Machine A": {
        "planned": 5000,
        "actual": 4700
    },

    "Machine B": {
        "planned": 4500,
        "actual": 3900
    },

    "Machine C": {
        "planned": 6000,
        "actual": 5700
    }
}

for machine, data in machines.items():

    planned = data["planned"]
    actual = data["actual"]

    efficiency = (actual / planned) * 100
    loss = planned - actual

    print(machine)
    print("Planned Production:", planned)
    print("Actual Production:", actual)
    print("Efficiency:", round(efficiency, 2), "%")
    print("Production Loss:", loss)

    if efficiency >= 90:
        print("Performance: Excellent")
    elif efficiency >= 75:
        print("Performance: Good")
    else:
        print("Performance: Poor")

    print("-------------------------")
```

```python
#Student Placement Prediction
students = {
    "Amit": {
        "cgpa": 8.4,
        "skills": 4,
        "internship": 1,
        "projects": 3
    },

    "Priya": {
        "cgpa": 9.1,
        "skills": 5,
        "internship": 2,
        "projects": 4
    },

    "Rahul": {
        "cgpa": 6.8,
        "skills": 2,
        "internship": 0,
        "projects": 1
    },

    "Sneha": {
        "cgpa": 8.7,
        "skills": 4,
        "internship": 1,
        "projects": 3
    }
}

for name, data in students.items():

    score = 0

    # CGPA score
    if data["cgpa"] >= 8.5:
        score += 40
    elif data["cgpa"] >= 7.5:
        score += 30
    else:
        score += 20

    # Skills
    score += data["skills"] * 5

    # Internship
    score += data["internship"] * 5

    # Projects
    score += data["projects"] * 5

    if score >= 75:
        prediction = "High Placement Chance"
    elif score >= 55:
        prediction = "Medium Placement Chance"
    else:
        prediction = "Low Placement Chance"

    print(name)
    print("Placement Score:", score)
    print("Prediction:", prediction)
    print("-------------------------")
```

```python
#Battery Health Analysis
batteries = {
    "Battery A": {"capacity": 95, "cycles": 320},
    "Battery B": {"capacity": 78, "cycles": 650},
    "Battery C": {"capacity": 91, "cycles": 410},
    "Battery D": {"capacity": 65, "cycles": 850}
}


for name, data in batteries.items():
    capacity = data["capacity"]
    cycles = data["cycles"]


    if capacity >= 90 and cycles < 500:
        status = "Excellent"
    elif capacity >= 75:
        status = "Good"
    else:
        status = "Replace Soon"


    print(name, ":", status)
    print("Capacity:", capacity, "%")
    print("Charge Cycles:", cycles)
    print()
```

```python
#Industrial Temperature Analysis
temperatures = [72, 75, 79, 84, 88, 91, 86, 78]


average = sum(temperatures) / len(temperatures)
maximum = max(temperatures)
minimum = min(temperatures)


print("Average Temperature:", round(average, 2), "°C")
print("Maximum Temperature:", maximum, "°C")
print("Minimum Temperature:", minimum, "°C")


if maximum > 90:
    print("Warning: Critical Temperature Detected")
else:
    print("Temperature is within safe range")
```

```python
#Vehicle Fuel Efficiency
vehicles = {
    "Vehicle A": {"distance": 450, "fuel": 30},
    "Vehicle B": {"distance": 520, "fuel": 40},
    "Vehicle C": {"distance": 600, "fuel": 35}
}
for vehicle, data in vehicles.items():
    mileage = data["distance"] / data["fuel"]


    print(vehicle, "Mileage:", round(mileage, 2), "km/l")
```

```python
#Air Quality Monitoring
air_quality = {
    "Station A": 45,
    "Station B": 85,
    "Station C": 155,
    "Station D": 220
}


for station, aqi in air_quality.items():


    if aqi <= 50:
        status = "Good"
    elif aqi <= 100:
        status = "Moderate"
    elif aqi <= 200:
        status = "Unhealthy"
    else:
        status = "Very Unhealthy"


    print(station, "AQI:", aqi, "->", status)
```
