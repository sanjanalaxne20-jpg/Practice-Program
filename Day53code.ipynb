```python
#Plastic Waste Percentage
total = float(input("Enter total waste (kg): "))
plastic = float(input("Enter plastic waste (kg): "))

percentage = (plastic / total) * 100

print("Plastic Waste Percentage =", round(percentage, 2), "%")
```

```python
#Solar Energy Generation
capacity = float(input("Enter solar panel capacity (kW): "))
sun_hours = float(input("Enter sunlight hours: "))

energy = capacity * sun_hours

print("Energy Generated =", energy, "kWh")
```

```python
#Rainwater Harvesting
rainfall = float(input("Enter rainfall (mm): "))
area = float(input("Enter roof area (square meters): "))

water = rainfall * area

print("Potential Rainwater =", water, "liters")
```

```python
#Tree Plantation Survival Rate
planted = int(input("Enter trees planted: "))
survived = int(input("Enter trees survived: "))

rate = (survived / planted) * 100

print("Tree Survival Rate =", round(rate, 2), "%")
```

```python
#Industrial Wastewater Treatment Efficiency
before = float(input("Enter pollutant concentration before treatment: "))
after = float(input("Enter pollutant concentration after treatment: "))

efficiency = ((before - after) / before) * 100

print("Treatment Efficiency =", round(efficiency, 2), "%")
```

```python
#Compost Production
organic_waste = float(input("Enter organic waste (kg): "))
conversion_rate = float(input("Enter compost conversion rate (%): "))

compost = organic_waste * conversion_rate / 100

print("Compost Produced =", compost, "kg")
```

```python
#Noise Pollution Check
noise = float(input("Enter noise level (dB): "))

if noise <= 50:
    print("Noise Level: Low")
elif noise <= 70:
    print("Noise Level: Moderate")
else:
    print("Noise Pollution: High")
```

```python
#Landfill Waste Calculation
total_waste = float(input("Enter total waste (tons): "))
recycled = float(input("Enter recycled waste (tons): "))
composted = float(input("Enter composted waste (tons): "))

landfill = total_waste - recycled - composted

print("Waste Sent to Landfill =", landfill, "tons")
```

```python
#Carbon Footprint Comparison
previous = float(input("Enter previous carbon footprint (kg): "))
current = float(input("Enter current carbon footprint (kg): "))

change = previous - current

if change > 0:
    print("Carbon Footprint Reduced by", change, "kg")
elif change < 0:
    print("Carbon Footprint Increased by", abs(change), "kg")
else:
    print("No Change")
```

```python
#Biodiversity Species Count
species = int(input("Enter total species observed: "))
endangered = int(input("Enter endangered species: "))

percentage = (endangered / species) * 100

print("Endangered Species Percentage =", round(percentage, 2), "%")
```
