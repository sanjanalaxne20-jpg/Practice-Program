```python
#Banking – EMI Calculation
#Problem: A customer pays a fixed EMI every month. Find the total amount paid.

emi = float(input("Enter monthly EMI: "))
months = int(input("Enter number of months: "))

total = emi * months

print("Total Amount Paid = ₹", total)
```

```python
#Hospital – BMI Calculator
#Problem: Calculate a patient's BMI.

weight = float(input("Enter weight (kg): "))
height = float(input("Enter height (m): "))

bmi = weight / (height ** 2)

print("BMI =", round(bmi, 2))
```

```python
#School – Grade Calculator
#Problem: Assign grades based on marks.
marks = int(input("Enter marks: "))

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

```python
#Agriculture – Crop Revenue
#Problem: Calculate total income from crop sales.

quantity = float(input("Enter crop quantity (kg): "))
price = float(input("Enter price per kg: "))

income = quantity * price

print("Total Income = ₹", income)
```

```python
#E-commerce – Discount Calculator
#Problem: Calculate the final bill after discount.

amount = float(input("Enter shopping amount: "))
discount = float(input("Enter discount (%): "))

final = amount - (amount * discount / 100)

print("Final Amount = ₹", final)
```

```python
#Hotel – Room Booking
#Problem: Find the total room charge.

days = int(input("Enter number of days: "))
rent = float(input("Enter room rent per day: "))

bill = days * rent

print("Total Bill = ₹", bill)
```

```python
#Library – Fine Calculation

#Problem: Calculate the fine for late book return.

days = int(input("Late days: "))

fine = days * 10

print("Fine = ₹", fine)
```

```python
#Transport – Bus Fare

#Problem: Calculate total fare for passengers.

passengers = int(input("Enter number of passengers: "))
fare = float(input("Fare per passenger: "))

total = passengers * fare

print("Total Fare = ₹", total)
```

```python
#Manufacturing – Salary

#Problem: Calculate workers' wages.

hours = int(input("Hours worked: "))
rate = float(input("Rate per hour: "))

salary = hours * rate

print("Salary = ₹", salary)
```

```python
#Sports – Average Score

#Problem: Calculate the average score of a player.

matches = int(input("Enter matches played: "))
runs = int(input("Enter total runs: "))

average = runs / matches

print("Average Score =", average)
```
