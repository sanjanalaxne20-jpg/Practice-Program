```python
#Check Buzz Number
def buzz(num):
    if num % 7 == 0 or num % 10 == 7:
        print("Buzz Number")
    else:
        print("Not Buzz Number")

n = int(input("Enter number: "))
buzz(n)
```

```python
#Check Duck Number
def duck(num):
    if "0" in str(num)[1:]:
        print("Duck Number")
    else:
        print("Not Duck Number")

n = int(input("Enter number: "))
duck(n)
```

```python
#Find Sum of Even Numbers in List
def even_sum(lst):
    total = 0
    for i in lst:
        if i % 2 == 0:
            total += i
    print(total)

l = list(map(int, input("Enter numbers: ").split()))
even_sum(l)
```

```python
#Find Sum of Odd Numbers
def odd_sum(lst):
    total = 0
    for i in lst:
        if i % 2 != 0:
            total += i
    print(total)

l = list(map(int, input("Enter numbers: ").split()))
odd_sum(l)
```

```python
#Count Even Numbers
def even_count(lst):
    count = 0
    for i in lst:
        if i % 2 == 0:
            count += 1
    print(count)

l = list(map(int, input("Enter numbers: ").split()))
even_count(l)
```

```python
#Count Odd Numbers
def odd_count(lst):
    count = 0
    for i in lst:
        if i % 2 != 0:
            count += 1
    print(count)

l = list(map(int, input("Enter numbers: ").split()))
odd_count(l)
```

```python
#Find Second Largest Number
def second_largest(lst):
    lst.sort()
    print(lst[-2])

l = list(map(int, input("Enter numbers: ").split()))
second_largest(l)
```

```python
#Remove Duplicates
def remove_duplicate(lst):
    print(list(set(lst)))

l = list(map(int, input("Enter numbers: ").split()))
remove_duplicate(l)
```

```python
#Count Words
def word_count(text):
    print(len(text.split()))

s = input("Enter sentence: ")
word_count(s)
```
