```python
#Waste Segregation
plastic = float(input("Plastic waste (kg): "))
organic = float(input("Organic waste (kg): "))
metal = float(input("Metal waste (kg): "))

if plastic >= organic and plastic >= metal:
    print("Highest Waste Type: Plastic")
elif organic >= plastic and organic >= metal:
    print("Highest Waste Type: Organic")
else:
    print("Highest Waste Type: Metal")
```

```python
#Water Saving Percentage
normal_usage = float(input("Enter normal water usage (liters): "))
new_usage = float(input("Enter new water usage (liters): "))

saving = ((normal_usage - new_usage) / normal_usage) * 100

print("Water Saved =", round(saving, 2), "%")
```

```python
#Energy Saving
old_energy = float(input("Enter previous energy consumption (kWh): "))
new_energy = float(input("Enter current energy consumption (kWh): "))

saving = old_energy - new_energy

print("Energy Saved =", saving, "kWh")
```

```python
#Environmental Monitoring – Average AQI
day1 = int(input("Enter AQI for Day 1: "))
day2 = int(input("Enter AQI for Day 2: "))
day3 = int(input("Enter AQI for Day 3: "))

average = (day1 + day2 + day3) / 3

print("Average AQI =", round(average, 2))
```

```python
#Environmental Compliance Check
emission = float(input("Enter measured emission: "))
limit = float(input("Enter permitted emission limit: "))

if emission <= limit:
    print("Status: Compliant")
else:
    print("Status: Non-Compliant")
```

```python
#CO₂ Reduction Target
current = float(input("Enter current CO2 emission (tons): "))
target_reduction = float(input("Enter reduction target (%): "))

reduction = current * target_reduction / 100
new_emission = current - reduction

print("CO2 to be Reduced =", reduction, "tons")
print("Target Emission =", new_emission, "tons")
```

```python
#Waste Collection Efficiency
scheduled = int(input("Enter scheduled collections: "))
completed = int(input("Enter completed collections: "))

efficiency = (completed / scheduled) * 100

print("Collection Efficiency =", round(efficiency, 2), "%")
```

```python
#Waste Diversion Rate
total = float(input("Enter total waste generated (tons): "))
landfill = float(input("Enter waste sent to landfill (tons): "))

diverted = total - landfill
rate = (diverted / total) * 100

print("Waste Diverted =", diverted, "tons")
print("Diversion Rate =", round(rate, 2), "%")
```

```python
#Wastewater Flow Rate
volume = float(input("Enter wastewater volume (liters): "))
time = float(input("Enter treatment time (hours): "))

flow_rate = volume / time

print("Average Flow Rate =", flow_rate, "liters/hour")
```

```python
#BOD Removal Efficiency
initial = float(input("Enter initial BOD (mg/L): "))
final = float(input("Enter final BOD (mg/L): "))

removal = ((initial - final) / initial) * 100

print("BOD Removal Efficiency =", round(removal, 2), "%")
```
