```python
#Electricity Power Cost
units = float(input("Units consumed: "))
rate = float(input("Rate per unit: "))

amount = units * rate

print("Electricity Cost =", amount)
```

```python
#Electricity Meter Reading
previous = int(input("Enter previous reading: "))
current = int(input("Enter current reading: "))

units = current - previous

print("Units Consumed =", units)
```

```python
#Train Ticket Fare
passengers = int(input("Enter number of passengers: "))
fare = float(input("Enter fare per passenger: "))

total = passengers * fare

print("Total Fare = ₹", total)
```

```python
#Newspaper Subscription
months = int(input("Enter number of months: "))
monthly_charge = float(input("Enter monthly charge: "))

bill = months * monthly_charge

print("Subscription Amount = ₹", bill)
```

```python
#Ice Cream Shop Bill
cones = int(input("Enter number of cones: "))
price = 40

bill = cones * price

print("Total Bill = ₹", bill)
```

```python
#Laptop Battery Backup
battery = int(input("Enter battery percentage: "))

hours = battery / 20

print("Estimated Backup =", hours, "hours")
```

```python
#Office ID Card Fee
employees = int(input("Enter number of employees: "))
fee = 150

total = employees * fee

print("Total ID Card Fee = ₹", total)
```

```python
#Water Bottle Stock
stock = int(input("Enter total bottles: "))
sold = int(input("Enter bottles sold: "))

remaining = stock - sold

print("Remaining Bottles =", remaining)
```

```python
#Hotel Food Bill
food = float(input("Enter food bill: "))
gst = food * 0.05

print("GST = ₹", gst)
print("Total Bill = ₹", food + gst)
```

```python
#Daily Salary
days = int(input("Enter working days: "))
salary_per_day = float(input("Enter salary per day: "))

salary = days * salary_per_day

print("Salary = ₹", salary)
```
