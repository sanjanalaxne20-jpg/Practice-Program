```python
#Employee ID Generator
name = input("Enter Name: ")
number = input("Enter Employee Number: ")

print("Employee ID =", name[:3].upper() + number)
```

```python
#Water Bill with Minimum Charge
units = int(input("Enter Water Units: "))

bill = max(units * 4, 100)

print("Water Bill = ₹", bill)
```

```python
#Library Membership
age = int(input("Enter Age: "))

if age >= 60:
    print("Senior Citizen Membership")
else:
    print("Regular Membership")
```

```python
#Salary Tax Calculator
salary = float(input("Enter annual salary: "))

if salary > 1000000:
    tax = salary * 0.20
elif salary > 500000:
    tax = salary * 0.10
else:
    tax = 0

print("Tax = ₹", tax)
print("Net Salary = ₹", salary - tax)
```

```python
#Package Delivery Status
status = input("Enter package status: ")

if status == "shipped":
    print("Package is on the way")
elif status == "delivered":
    print("Package delivered")
elif status == "cancelled":
    print("Order cancelled")
else:
    print("Unknown status")
```

```python
#Restaurant Table Booking
tables = int(input("Enter total tables: "))
booked = int(input("Enter booked tables: "))

available = tables - booked

if available > 0:
    print("Tables Available =", available)
else:
    print("No Table Available")
```

```python
#Student Attendance Warning
total_classes = int(input("Enter total classes: "))
present = int(input("Enter classes attended: "))

attendance = (present / total_classes) * 100

print("Attendance =", round(attendance, 2), "%")

if attendance < 75:
    print("Attendance Warning")
else:
    print("Attendance Satisfactory")
```

```python
#E-commerce Free Delivery
amount = float(input("Enter order amount: "))

if amount >= 999:
    print("Free Delivery")
else:
    print("Delivery Charge = ₹80")
```

```python
#Employee Performance Rating
rating = float(input("Enter performance rating out of 5: "))

if rating >= 4.5:
    print("Excellent Performance")
elif rating >= 3.5:
    print("Good Performance")
elif rating >= 2.5:
    print("Average Performance")
else:
    print("Needs Improvement")
```
